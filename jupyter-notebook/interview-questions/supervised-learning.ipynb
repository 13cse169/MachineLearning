```python
# 5. What is a feature and what is a target variable?

# 🧠 Simple idea
# Features (X) → input data (what you give to model)
# Target (y) → output (what you want to predict)

# 📌 Example
# Hours studied → feature
# Pass/Fail → target

# Features are the input variables used to make predictions, while the target variable is the output we want the model to predict.
```

```python
# 6. What is training data and testing data? Why do we split?

# 🧠 Simple idea

# We don’t want the model to just memorize data.

# Train data → used to learn
# Test data → used to check performance on unseen data

# We split data into training and testing sets to evaluate how well the model performs on unseen data and to avoid overfitting.
```

```python
# 7. What is overfitting?

# 🧠 Simple idea

# Model becomes too smart on training data, but fails on new data.

# 📌 Example
# Training accuracy = 100%
# Test accuracy = low

# Overfitting occurs when a model learns the training data too well, including noise, and performs poorly on new unseen data.
```

```python
# 8. What is underfitting?

# 🧠 Simple idea

# Model is too simple, doesn’t learn properly.

# 📌 Example
# Training accuracy = low
# Test accuracy = also low

# Underfitting happens when a model is too simple to capture the underlying pattern in the data, resulting in poor performance on both training and test data.
```

```python
# 9. How can you detect overfitting?

# 🧠 Simple idea

# Compare training vs testing performance

# ✅ Signs
# High training accuracy
# Low testing accuracy

# Overfitting can be detected when the model performs very well on training data but poorly on test data, indicating it has not generalized well.

# ✅ How to reduce overfitting (give any 2 in interview)
# Use more training data, so the model learns general patterns instead of memorizing.
# Reduce model complexity, for example using a smaller model or tuning parameters.

# 💡 You can also mention (if asked more)
# Use cross-validation
# Apply regularization (L1/L2)
# Prune decision trees
# Use dropout (in neural networks)

# 🧠 Easy way to remember
# Overfitting = “model memorized”
# Solution = “make it generalize”
```

```python
# 🧠 Quick memory trick
# Overfitting → “memorized”
# Underfitting → “didn’t learn”
```

```python
# 10. How does KNN (K-Nearest Neighbors) work in simple words?

# 🧠 Simple idea
# It looks at nearest data points
# Takes a majority vote (for classification)
# 📌 Example

# If K = 3:

# Check 3 nearest neighbors
# If 2 are “Pass” and 1 is “Fail” → result = Pass

# KNN works by finding the K nearest data points to a given input and predicting the output based on the majority class (for classification) or average value (for regression).
```

```python
# 11. What happens if K is too small?

# 🧠 Idea
# Model becomes too sensitive
# Can overfit

# If K is too small, the model becomes sensitive to noise and may overfit the data.
```

```python
# 12. What happens if K is too large?

# 🧠 Idea
# Model becomes too general
# May ignore important patterns

# If K is too large, the model becomes too generalized and may underfit the data.
```

```python
# 13. What is the difference between Linear Regression and Logistic Regression?

# 🧠 Key difference
# Linear Regression	        Logistic Regression
# Predicts numbers	        Predicts categories
# Output range (-∞ to +∞)	Output (0 to 1)

# Linear regression is used for predicting continuous values, while logistic regression is used for classification problems and outputs probabilities between 0 and 1.
```

```python
# 14. For classification, what is accuracy?

# 🧠 Simple idea

# Out of all predictions, how many were correct

# 📌 Formula

# Accuracy = Correct Predictions / Total Predictions

# Accuracy is the ratio of correctly predicted observations to the total observations.
```

```python
# 15. When is accuracy NOT a good metric?

# 🧠 Important concept (very common question)

# When data is imbalanced

# 📌 Example
# 95% people = No disease
# Model predicts always “No disease”
# Accuracy = 95% ❌ but useless model

# Accuracy is not a good metric when the dataset is imbalanced because it can give misleading results.
```

```python
# 16. What is precision and recall?

# 🧠 Simple understanding
# Precision → out of predicted positives, how many are correct
# Recall → out of actual positives, how many we found

# 📌 Example (Spam detection)
# Precision → how many predicted spam are actually spam
# Recall → how many actual spam we correctly detected

# Precision measures how many predicted positive cases are actually correct, while recall measures how many actual positive cases are correctly identified.
```

```python
# 17. For regression, name 2 evaluation metrics.

# Common ones:
    # MAE (Mean Absolute Error)
    # MSE (Mean Squared Error)
    # RMSE

# Common regression metrics include Mean Absolute Error and Mean Squared Error, which measure the difference between predicted and actual values.
```

```python
# 18. Why do we use GridSearchCV?

# 🧠 Idea

# Models have parameters (like K in KNN). We need the best ones.

# 📌 What it does
# Tries different combinations
# Uses cross-validation
# Picks best parameters automatically

# GridSearchCV is used to find the best combination of hyperparameters by testing multiple values and selecting the one that gives the best performance.
```

```python
# 19. What is cross-validation?

# 🧠 Idea

# Instead of one split, we test multiple splits

# 📌 Example (cv=3)
# Split data into 3 parts
# Train on 2, test on 1
# Repeat 3 times

# Cross-validation is a technique where the dataset is split into multiple parts to evaluate the model on different subsets, giving a more reliable performance estimate.
```

```python
# 20. What is the difference between fit() and predict()?

# 🧠 Idea
# fit() → learning
# predict() → using learned model

# The fit() method is used to train the model on data, while predict() is used to make predictions on new data.

# What will happen if you call predict() before fit()?

# If we call predict() before fit(), the model will throw an error because it has not been trained yet and does not know the learned parameters.
```

```python
# 22. What is bias vs variance tradeoff?

# 🧠 Simple idea
# Bias → model is too simple (underfitting)
# Variance → model is too complex (overfitting)
# 🎯 Goal

# Balance both

# Bias refers to error due to overly simple assumptions in the model, while variance refers to error due to the model being too sensitive to training data. The goal is to balance both to achieve good performance.
```

```python
# 23. Why do we scale data in KNN?

# 🧠 Key idea

# KNN uses distance

# If one feature is large (like salary) and another small (like age),
# 👉 large feature dominates

# 📌 Example
# Age: 20–60
# Salary: 10,000–1,00,000

# Distance becomes biased toward salary

# We scale data in KNN because it is a distance-based algorithm, and features with larger values can dominate the distance calculation.
```

```python
# 23. What happens if your dataset has missing values?

# 🧠 Options
# Remove rows
# Fill with mean/median/mode
# Use advanced methods

# Missing values can be handled by removing rows or columns, or by imputing values using mean, median, or mode depending on the data.
```
