```python
# 🚀 Machine Learning Interview Preparation

# 📌 What is Machine Learning?
# Machine Learning is a method where models learn patterns from data and make predictions without being explicitly programmed.

# 📌 Types of ML:
#         Supervised
#         Unsupervised
#         Reinforcement

# 📌 Overfitting vs Underfitting
# Overfitting happens when a model performs well on training data but poorly on test data. Underfitting happens when the model is too simple to capture patterns.

# 📌 Bias vs Variance
# High Bias → Underfitting
# High Variance → Overfitting

# 📌 What is Cross Validation?
# Technique to evaluate model performance by splitting data multiple times and averaging results.

# 📌 What is Regularization?

# Used to reduce overfitting.

# Types:

# L1 (Lasso)
# L2 (Ridge)
```

```python
# 🔷 2. Algorithm-Based Questions

# 📌 Linear Regression
# Used for regression
# Assumes linear relationship

# 📌 Logistic Regression
# Used for classification
# Uses sigmoid function

# 📌 Decision Tree
# Based on entropy / gini
# Easy to interpret

# 📌 Random Forest
# Ensemble of trees
# Reduces overfitting

# 📌 Support Vector Machine
# Finds optimal hyperplane
# Uses kernel trick

# 📌 K-Means Clustering
# Unsupervised
# Groups similar data

# 📌 Principal Component Analysis
# Reduces dimensions
# Keeps max variance
```

```python
# 🔷 3. Common Interview Questions

# Q1: Why do we split data?

# 👉 To evaluate model on unseen data

# Q2: What is feature scaling?

# 👉 Scaling features to same range for distance-based algorithms like

# K-Nearest Neighbors
# Support Vector Machine

# Q3: What is confusion matrix?
# 	          Pred Yes	Pred No
# Actual Yes	  TP	    FN
# Actual No	  FP	    TN

# Q4: Precision vs Recall
# Precision → correctness of positive predictions
# Recall → how many positives detected
```

```python
# 🔷 4. Coding Question (Very Common)

# 👉 Train a model and evaluate

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

X = [[1],[2],[3],[4],[5],[6]]
y = [0,0,0,1,1,1]

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2)

model = LogisticRegression()
model.fit(X_train,y_train)

y_pred = model.predict(X_test)

print(accuracy_score(y_test,y_pred))
```
