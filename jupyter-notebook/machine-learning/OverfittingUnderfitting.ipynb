```python
# 1. Overfitting vs Underfitting
```

```python
# Underfitting: Model is too simple -> cannot learn pattern properly.
    # Poor performance on:
        # Training Data
        # Test Data
```

```python
# Overfitting: Model memorizes trainging data insted of learning.
    # Performance:
        # Training Data (Very Good)
        # Test Data (Poor)

# Overfitting happens when a model performs very well on training data but poorly on unseen data because it learns noise instead of general patterns.
```

```python
# 2. Intution:
    # Underfitting: Data is complex, but model os too simple -> misses patterns.
    # Example: Using Linear Regression for a curved dataset

    # Overfitting: Model tries to fit every single point -> Learn noise
    # Example: Using a deep Decision Tree without limits
```

```python
# 3. Visual Understanding
# --------------------------------------------
# Case           Behavior
# --------------------------------------------
# Underfitting	Straight line for curved data
# Overfitting	Zig-zag line touching all points
# Best Fit	    Smooth curve
```

```python
# 4. Code Example:

# Step 1: Import Libraries

import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Step 2: Create Data

np.random.seed(0)

X = np.sort(5 * np.random.rand(40, 1), axis=0)
y = np.sin(X).ravel()

# Step 3: Underfitting (Simple Model)

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

plt.scatter(X, y)
plt.plot(X, y_pred)
plt.title("Underfitting")
plt.show()
```

```python
# Step 4: Overfitting (Complex Model)

poly = PolynomialFeatures(degree=15)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

y_pred = model.predict(X_poly)

plt.scatter(X, y)
plt.plot(X, y_pred)
plt.title("Overfitting")
plt.show()
```

```python
# Step 5: Good Fit

poly = PolynomialFeatures(degree=3)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

y_pred = model.predict(X_poly)

plt.scatter(X, y)
plt.plot(X, y_pred)
plt.title("Good Fit")
plt.show()
```

```python
# 5. How to Detect:
# --------------------------------------------------
# Case	         Training Error	Test Error
# --------------------------------------------------
# Underfitting	 High	        High
# Overfitting	 Low	        High
# Good Fit	     Low	        Low
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

np.random.seed(0)

X = np.sort(5 * np.random.rand(40, 1), axis=0)
y = np.sin(X).ravel()

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

plt.scatter(X, y)
plt.plot(X, y_pred)
plt.title("Underfitting")
plt.show()
```

```python
# Step 4: Overfitting (Complex Model)

poly = PolynomialFeatures(degree=1)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

y_pred = model.predict(X_poly)

plt.scatter(X, y)
plt.plot(X, y_pred)
plt.title("Overfitting: Degree=1")
plt.show()
```

```python
# Step 4: Overfitting (Complex Model)

poly = PolynomialFeatures(degree=5)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

y_pred = model.predict(X_poly)

plt.scatter(X, y)
plt.plot(X, y_pred)
plt.title("Overfitting: Degree=5")
plt.show()
```
