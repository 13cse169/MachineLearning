```python
# 🔷 Cross Validation:

# 📌 Cross Validation is a technique to check how well your model performs on unseen data by testing it multiple times on different splits.
# 📌 Cross Validation is a technique where data is split into multiple parts and the model is trained and tested multiple times to get a reliable performance estimate.

# 👉 Instead of splitting data once, we split it multiple times.
```

```python
# 🔷 1. Why Do We Need It?

# Problem with normal split: Train → 80% and Test → 20%
# 👉 Result depends on how the data was split
# 👉 Can give misleading accuracy

# ✅ Solution: Cross Validation

# We divide data into K parts (folds).

# Example: 5-Fold Cross Validation

# Fold 1 → Test, rest Train  
# Fold 2 → Test, rest Train  
# Fold 3 → Test, rest Train  
# Fold 4 → Test, rest Train  
# Fold 5 → Test, rest Train  

# Final result = average of all scores
```

```python
# 🔷 2. Types of Cross Validation
# 🔹 K-Fold (Most Common)
# Data split into K equal parts
# Default: K = 5 or 10
# 🔹 Stratified K-Fold

# Used for classification (keeps class balance)
```

```python
# 🔷 3. Code Example:

# Step 1: Import

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
import pandas as pd

# Step 2: Create Data

data = {
    "hours":[1,2,3,4,5,6],
    "pass":[0,0,0,1,1,1]
}

df = pd.DataFrame(data)

X = df[["hours"]]
y = df["pass"]

# Step 3: Apply Cross Validation

model = LogisticRegression()

scores = cross_val_score(model, X, y, cv=3)

print("Scores: ", scores)
print("Average Accuracy: ", scores.mean())
```

```python
# 🔷 Example Output
# Scores: [0.5, 1.0, 1.0]
# Average Accuracy: 0.83

# 👉 This is more reliable than a single split
```

```python
# 🔷 4. Visual Understanding

# Instead of: 1 test → result

# You get: 5 tests → average result

# 👉 More stable
# 👉 Less overfitting
```

```python
# 🔷 5. When to Use

# Use Cross Validation when:

# ✔ Dataset is small
# ✔ You want reliable accuracy
# ✔ You want to compare models
```
