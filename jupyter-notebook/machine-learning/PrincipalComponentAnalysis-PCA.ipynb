```python
# 🔷 Lesson — Principal Component Analysis (PCA)
# 📌 1. What is PCA?

# PCA (Principal Component Analysis) is used to:

# 👉 Reduce the number of features while keeping most of the important information.
```

```python
# 🧠 Simple Example

# Suppose you have data with:

# Height, Weight, Age, Salary, Experience

# 👉 Too many features → complex model

# PCA converts them into:

# PC1, PC2

# 👉 Fewer features, but still useful information
```

```python
# 🔷 2. Why Do We Need PCA?
# Problems without PCA:

# ❌ Too many features
# ❌ Slow models
# ❌ Overfitting
# ❌ Hard to visualize

# Benefits of PCA:

# ✔ Reduces dimensions
# ✔ Faster training
# ✔ Removes noise
# ✔ Helps visualization
```

```python
# 🔷 3. Key Idea (Very Important)

# PCA finds new axes (directions) where:

# 👉 Maximum variance (information) is preserved

# These are called:

# 👉 Principal Components

# 📌 First Component (PC1)

# Captures maximum variance

# 📌 Second Component (PC2)

# Captures remaining variance
```

```python
# 🔷 4. Visual Intuition
# Original data → tilted cloud

# PCA rotates axes → captures max spread
```

```python
# 🔷 5. Code Implementation (Step-by-Step)

import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

data = {
    "income":[20000,30000,50000,80000,100000],
    "spending":[20,30,50,80,90]
}

df = pd.DataFrame(data)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

pca = PCA(n_components=1)

X_pca = pca.fit_transform(X_scaled)

X_pca
```

```python
# 🔷 6. Explained Variance: This tells how much information is kept.

pca.explained_variance_ratio_

# [0.95] 👉 95% information retained
```

```python
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X_pca[:,0], X_pca[:,1])

plt.xlabel("PC1")
plt.ylabel("PC2")

plt.show()
```

```python
# 🔷 9. Combine PCA with K-Means

# Very powerful combo:

# 👉 Use PCA to reduce features
# 👉 Then apply K-Means Clustering
```

```python
from sklearn.cluster import KMeans

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

model = KMeans(n_clusters=2)

clusters = model.fit_predict(X_pca)
```

```python
# 🔷 10. When to Use PCA

# Use PCA when:

# ✔ Dataset has many features
# ✔ Features are correlated
# ✔ Need faster model
# ✔ Want visualization
```

```python
# 🔷 11. Limitations

# ✘ Hard to interpret new features
# ✘ Some information loss
# ✘ Not always needed
```
