```python
# 🔷 Hyperparameter Tuning — GridSearchCV

# 📌 Hyperparameter tuning means finding the best settings for a model to improve performance.

# 👉 You don’t change the algorithm
# 👉 You change its configuration

# GridSearchCV is used to find the best hyperparameters by testing all combinations using cross-validation and selecting the one with the highest performance.
```

```python
# 🔷 1. What are Hyperparameters?

    # These are values you set before training.

        # Algorithm	                Hyperparameters
        # K-Nearest Neighbors	    K (number of neighbors)
        # Random Forest	            n_estimators, max_depth
        # Support Vector Machine	C, kernel, gamma
```

```python
# 🔷 2. Problem Without Tuning

    # Example:
    
    # K = 3 → accuracy = 70%
    # K = 7 → accuracy = 85%
    
    # 👉 Which one is best?
    # 👉 We don’t know manually
```

```python
# 🔷 3. Solution — Grid Search

    # Try all combinations and pick the best.
    
    # Example:
    
    # K = [3,5,7]
    # weights = ["uniform","distance"]
    
    # Total combinations:
    
    # 3 × 2 = 6 models tested
```

```python
# 🔷 4. Code Example (Step-by-Step)

# Step 1: Import
from sklearn.model_selection import GridSearchCV
from sklearn.neighbors import KNeighborsClassifier
import pandas as pd

# Step 2: Data
data = {
    "hours":[1,2,3,4,5,6],
    "pass":[0,0,0,1,1,1]
}

df = pd.DataFrame(data)

X = df[["hours"]]
y = df["pass"]

# Step 3: Define Model
model = KNeighborsClassifier()

# Step 4: Define Parameters
param_grid = {
    "n_neighbors": [1,3,5],
    "weights": ["uniform", "distance"]
}

# Step 5: Apply GridSearch
grid = GridSearchCV(model, param_grid, cv=3)
grid.fit(X, y)

# Step 6: Best Result
print("Best Params: ", grid.best_params_)
print("Best Score: ", grid.best_score_)
```

```python
# 🔷 5. Why It’s Powerful

    # ✔ Finds best model automatically
    # ✔ Uses cross validation internally
    # ✔ Improves accuracy

# 🔷 6. Important Tip

    # Use with:
    
    # Random Forest
    # Support Vector Machine
    
    # These models depend heavily on tuning.

# 🔷 7. Faster Alternative

    # 👉 RandomizedSearchCV
    
    # Tries random combinations
    # Faster than GridSearch
    # Useful for large datasets

# 🔷 8. Common Mistakes

    # ❌ Using too many parameters (very slow)
    # ❌ Not using cross-validation
    # ❌ Tuning before understanding model
```
