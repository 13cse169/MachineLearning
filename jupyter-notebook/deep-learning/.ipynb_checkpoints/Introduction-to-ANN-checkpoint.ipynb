```python
# Step 1: Import

import numpy as np
from tensorflow import keras
from tensorflow.keras import layers
```

```python
# Step 2: Data

X = np.array([[0],[1],[2],[3],[4],[5]])
y = np.array([0,0,0,1,1,1])
```

```python
# Step 3: Build Model

model = keras.Sequential([
    layers.Dense(1, activation="sigmoid")
])
```

```python
# 🔷 6. Code Implementation Using TensorFlow
```

```python
# Step 1 — Import
import numpy as np
from tensorflow import keras
from tensorflow.keras import layers
```

```python
# Step 2 — Dataset
X = np.array([
    [1,10],
    [2,20],
    [3,30],
    [4,40],
    [5,50],
    [6,60]
])

y = np.array([0,0,0,1,1,1])
```

```python
# 🔷 7. Build Deep Neural Network

model = keras.Sequential([
    keras.Input(shape=(2,)),
    
    # layers.Dense(8, activation="relu", input_shape=(2,)),
    
    layers.Dense(8, activation="relu"),

    layers.Dense(4, activation="relu"),

    layers.Dense(1, activation="sigmoid")
])
```

```python
# 🔷 8. Understanding This Architecture

    # 🔹 First Hidden Layer - Dense(8)

        # Means: 👉 8 neurons in hidden layer

    # 🔹 Second Hidden Layer - Dense(4)

        # Means: 👉 4 neurons

    # 🔹 Output Layer - Dense(1, activation="sigmoid")

    # Binary output:

        # 0 → No
        # 1 → Yes
```

```python
# 🔷 9. Compile Model

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
# 🔷 10. Train Model
model.fit(X, y, epochs=100)
```

```python
# 🔷 11. Prediction
prediction = model.predict(np.array([[5,45]]))

print(prediction)
```
